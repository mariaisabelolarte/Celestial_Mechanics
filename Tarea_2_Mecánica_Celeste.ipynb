Cálculo de Asención Recta y Declinación de Apophis usando sus coordenadas cartesianas

Se define una función que convierte grados al formato de grados, minutos y segundos para una mejor comprensión de los resultados

```python
def conversion_angulos(angulo):
    grados = int(angulo)
    minutos = int((abs(angulo - grados)) * 60)
    segundos = round(((abs(angulo - grados)) * 60 - minutos) * 60, 2)

    if angulo < 0:
        grados = -abs(grados)

    return grados, minutos, segundos
```

Se obtienen las coordenadas cartesianas de Apophis en el momento de la aproximación y se convierten en coordenadas eclípticas mediante las expresiones:


$\tan{\lambda} = \frac{y}{x}$

$\sin{\beta} = \frac{z}{\mid r \mid}$

```python
coor_apophis = rs_apophis_refi[indice, :]
```

Conversión a coordenadas eclípticas

```python
longitud = np.arctan(coor_apophis[1] / coor_apophis[0])
longitud = longitud * 180 / np.pi

grados, minutos, segundos = conversion_angulos(longitud)
print(f"Longitud Eclíptica = {grados}° {minutos}' {segundos}\"")

beta = np.arcsin(coor_apophis[2] / np.linalg.norm(coor_apophis))
beta = beta * 180 / np.pi

grados, minutos, segundos = conversion_angulos(beta)
print(f"Latitud Eclíptica = {grados}° {minutos}' {segundos}\"")
```

Para convertir a coordenadas ecuatoriales absolutas, utilizamos las fórmulas de conversión encontradas en el libro *Elementos de Astronomía de Posición* de Héctor Portilla:

$\sin{\delta} = \sin{\beta} \cos{\epsilon} + \cos{\beta}\sin{\epsilon}\sin{\lambda}$

$\tan{\alpha} = \frac{-\sin{\beta}\sin{\epsilon}+\cos{\beta}\cos{\epsilon}\sin{\lambda}}{\sin{\lambda}\sin{\beta}}$

Donde $\epsilon$ corresponde a la oblicuidad de la eclíptica. Además se tiene que:

$\alpha$ es de la forma $\tan{\alpha}= \tan^{-1}{\left( \frac{p}{q} \right)}$, y el ángulo verdadero se encuentra según las reglas:

Si $p \cdot q < 0$ y $q<0$ entonces $\alpha = \alpha + 180$

Si $p \cdot q < 0$ y $q>0$ entonces $\alpha = \alpha + 360$

Si $p+q<0$ entonces $\alpha = \alpha + 180$

Teniendo en cuenta las fórmulas anteriores, creamos una función que convierte coordenadas eclípticas en coordenadas ecuatoriales absolutas.

```python
def coordenadas_ecuatoriales(long_eclip, lat_eclip):
    epsilon = 23.44 # oblicuidad de la eclíptica

    # radianes
    epsilon = epsilon * np.pi / 180
    long_eclip = long_eclip * np.pi / 180
    lat_eclip = lat_eclip * np.pi / 180

    # conversión
    sin_delta = (np.cos(epsilon) * np.sin(lat_eclip) + np.sin(epsilon) \
                 * np.cos(lat_eclip) * np.sin(long_eclip))
    delta = np.arcsin(sin_delta)

    p = (- np.sin(lat_eclip) * np.sin(epsilon) + np.cos(lat_eclip) \
         * np.cos(epsilon) * np.sin(long_eclip))
    q = np.cos(lat_eclip) * np.cos(long_eclip)

    tan_alpha = p / q
    alpha = np.arctan(tan_alpha)

    if p * q < 0 and q < 0:
        alpha = alpha + np.pi
    elif p * q < 0 and q > 0:
        alpha = alpha + 2*np.pi
    elif p + q < 0:
        alpha = alpha + np.pi
    else:
        alpha = alpha

    # grados
    delta = delta * 180 / np.pi
    alpha = alpha * 180 / np.pi

    return alpha, delta
```

Se aplica la función de conversión de coordenadas y se obtienen las coordenadas ecuatoriales absolutas de Apophis en el momento de la aproximación

```python
alpha, delta = coordenadas_ecuatoriales(longitud, beta)

horas = int(alpha / 15)
minutos = int((alpha / 15 - horas) * 60)
segundos = round((((alpha / 15 - horas) * 60) - minutos) * 60, 2)

print(f"Ascensión Recta = {horas} hora {minutos} min {segundos} sec")

grados, minutos, segundos = conversion_angulos(delta)
print(f"Declinación = {grados}° {minutos}' {segundos}\"")
```

Utilizando el software Stellarium se encuentra que dichas coordenadas ecuatoriales absolutas corresponden a la constelación de Pisces. La noche del 14 de abril de 2029 dicha constelación no será visible en Colombia, pues el Sol se encontrará en dicha constelación, dificultando su observación durante la noche.

# **2. Cambio de elementos orbitales después de la aproximación**

Nuevamente integramos con Rebound la trayectoria de Apophis, tomando como  momento inicial el de la aproximación obtenido en la primera simulación de Rebound realizada anteriormente. Para cada paso, obtendremos los elementos orbitales

Creamos la simulación en Rebound

```python
sim2 = rb.Simulation()
sim2.units = 'Msun', 'au', 'yr'
```

Agregamos a la simulación los cuerpos y definimos como fecha inicial el momento de aproximación calculado en el punto 1 (14 de abril de 2029 a las 9 am en UTC).

```python
for i, id_body in enumerate(('10', '199', '299', '399', '499', '599',
                             '699', '799', '899', 'Apophis')):
  sim2.add(id_body, date = '2029-04-14 09:00:00')
```

Se define el tiempo total de integración (1 año) y el número de veces que se integrará

```python
Nt = 1000
t_total = 1
ts = np.linspace(0, t_total, Nt)
```

Creamos un cubo de datos para guardar los elementos orbitales en cada iteración

```python
elementos = np.zeros((Nt, 6))
```

Integramos y obtenemos los elementos orbitales. Guardamos los resultados en el cubo de datos

```python
for n, t in enumerate(ts):
    sim2.integrate(t)
    orb = sim2.particles[9].orbit(primary = sim2.particles[0])
    elementos[n] = [orb.a, orb.e, orb.inc, orb.omega, orb.Omega, orb.M]
```

Creamos un array con las fechas de integración para visualizar mejor las siguientes gráficas. Usamos el mismo código utilizado en el punto 1

```python
start = pd.Timestamp("2029-04-14 09:00:00")
end = start + pd.DateOffset(years = t_total)
fechas = pd.date_range(start = start, end = end, periods = Nt)
```

Gráficas con los cambios en los elementos orbitales en función del tiempo

```python
fig, axs = plt.subplots(3, 2, figsize = (10, 8))

axs[0, 0].plot(fechas, elementos[:, 0], color = 'blue')
axs[0, 0].set_title('$a$ vs $t$')
axs[0, 0].tick_params(axis = 'x', rotation = 45)
axs[0, 0].set_ylabel('$a$ [AU]')
axs[0, 0].set_xlabel('Fecha')

axs[0, 1].plot(fechas, elementos[:, 1], color = 'lime')
axs[0, 1].set_title('$e$ vs $t$')
axs[0, 1].tick_params(axis = 'x', rotation = 45)
axs[0, 1].set_ylabel('$e$')
axs[0, 1].set_xlabel('Fecha')

axs[1, 0].plot(fechas, elementos[:, 2], color = 'magenta')
axs[1, 0].set_title('$i$ vs $t$')
axs[1, 0].tick_params(axis = 'x', rotation = 45)
axs[1, 0].set_ylabel('$i$ [rad]')
axs[1, 0].set_xlabel('Fecha')

axs[1, 1].plot(fechas, elementos[:, 3], color = 'orange')
axs[1, 1].set_title('$\omega$ vs $t$')
axs[1, 1].tick_params(axis = 'x', rotation = 45)
axs[1, 1].set_ylabel('$\omega$ [rad]')
axs[1, 1].set_xlabel('Fecha')

axs[2, 0].plot(fechas, elementos[:, 4], color = 'darkviolet')
axs[2, 0].set_title('$\Omega$ vs $t$')
axs[2, 0].tick_params(axis = 'x', rotation = 45)
axs[2, 0].set_ylabel('$\Omega$ [rad]')
axs[2, 0].set_xlabel('Fecha')

axs[2, 1].plot(fechas, elementos[:, 5], color = 'turquoise')
axs[2, 1].set_title('$M$ vs $t$')
axs[2, 1].tick_params(axis = 'x', rotation = 45)
axs[2, 1].set_ylabel('$M$ [rad]')
axs[2, 1].set_xlabel('Fecha')

fig.suptitle('Elementos Orbitales de Apophis después de su '
             'Aproximación a la Tierra', fontsize=16)
plt.tight_layout()
plt.show()
```

Se observa que, luego de su aproximación con la Tierra, los elementos orbitales de Apophis se estabilizan. Por lo tanto, para determinar cuánto cambiaron, tomamos los valores de los elementos al iniciar la integración (momento de aproximación) y al final de la integración. Esto nos dará una idea de cómo cambiaron $a, e, i, \omega , \Omega$.

En el caso de $M$, observamos que está aumentando constantemente, lo cual concuerda con lo esperado, pues este elemento orbital depende del tiempo y va siempre cambiando a medida que Apophis avanza por su órbita.

```python
cambio_a = elementos[-1, 0] - elementos[0, 0]
cambio_e = elementos[-1, 1] - elementos[0, 1]
cambio_i = elementos[-1, 2] - elementos[0, 2]
cambio_omega = elementos[-1, 3] - elementos[0, 3]
cambio_Omega = elementos[-1, 4] - elementos[0, 4]
```

```python
print(f"Cambio en a: {np.round(cambio_a, 4)} UA")
print(f"Cambio en e: {np.round(cambio_e, 4)}")
print(f"Cambio en i: {np.round(cambio_i, 4)} rad")
print(f"Cambio en omega: {np.round(cambio_omega, 4)} rad")
print(f"Cambio en Omega: {np.round(cambio_Omega, 4)} rad")
```

Dichos cambios pueden interpretarse así:

La disminución de $a$ puede deberse a un impulso gravitacional de la Tierra sobre Apophis durante la aproximación, haciendo que cambie el semieje mayor y posteriormente se mantenga constante (como lo indica la teoría de los dos cuerpos)

La disminución de $e$ luego del sobrevuelo de Apophis cerca de la Tierra indica que se dio un cambio de la forma de la órbita de Apophis debido a la perturbación terrestre, haciendola un poco más circular. Luego de dicho cambio, la excentricidad se estabiliza y permanece constante.

La disminución de $i$ puede indicar que una componente vertical del impulso gravitacional de la tierra disminuyó la inclinación de la órbita de Apophis. Nuevamente, después de la aproximación la inclinación se estabiliza y permanece constante.

El cambio de $\omega$ se observa como una disminución con una pequeña oscilación. Puede interpretarse como una precesión del perihelio inducida por la perturbación. Luego del acercamiento, el argumento del perihelio se estabiliza y permanece constante.

El cambio de $\Omega$ indica una rotación del plano orbital respecto a la referencia. Luego de la aproximación, la longitud del nodo ascendente se estabiliza y permanece constante.

Como se explicó anteriormente, la evolución de $M$ es esperada, pues es un elemento orbital que varía de forma constante con le tiempo (como se observa en la gráfica). En las gráficas no se observa un cambio en el comportamiento normal de la anomalía media, y despúes de la aproximación sigue cambiando de la manera esperada.

# **3. Fecha y distancia de aproximación mediante el problema de los dos cuerpos**

La consulta a horizons nos da los parámetros: $a, e, I, \Omega, \omega, f, M, P, to$. Guardamos en variables los parámetros orbitales que vamos a utilizar. Además convertimos el semieje mayor a unidades astronómicas y los ángulos a radianes

```python
a = parametros[0] / UL
e = parametros[1]
incl = parametros[2] * np.pi / 180 % (2 * np.pi)
Omega = parametros[3] * np.pi / 180 % (2 * np.pi)
omega = parametros[4] * np.pi / 180 % (2 * np.pi)
M = parametros[6] * np.pi / 180 % (2 * np.pi)
```

A partir de los elementos orbitales calculamos el semilatus rectum de la órbita de Apophis

Calcularemos la posición de Apophis cada 6 horas durante un poco más de su periodo orbital de 323.6 días, por lo que el número de iteraciones será:

```python
pasos = int(327 * 4)
```

Creamos arrays para guardar los valores de las anomalías media, excéntrica y verdadera en cada iteración. Guardamos la anomalía media inicial.

```python
Ms = np.zeros(pasos)
Es = np.zeros(pasos)
fs = np.zeros(pasos)
```

Definimos el valor inicial de la anomalía media como el consultado de horizons

```python
Ms[0] = M
```

Definimos una función que calcule la anomalía verdadera a partir de la anomalía excéntrica, según la fórmula:

$tan\frac{f}{2} = \sqrt\frac{1+e}{1-e} tan\frac{E}{2}$

```python
def anomalia_verdadera(E):
  tan_f_2 = np.sqrt((1 + e)/(1 - e)) * np.tan(E / 2)
  return 2 * np.arctan(tan_f_2)
```

Calulamos la velocidad angular promedio de Apophis en radianes por hora ($n$)

```python
n = 2 * np.pi / (323.6 * 24)
```

A partir de la velocidad angular promedio $n$, calculamos la anomalía media $M$ en cada iteración y a partir de ella calculamos la anomalía excéntrica $E$ en cada iteración mediante la ecuación de Kepler y el método de Newton:

$E-esin(E)=n(t-t_0)$

Luego de tener la anomalía excéntrica, calculamos la anomalía verdadera $f$ empleando la función definida anteriormente.

Guardamos los resultados en los arreglos creados anteriormente.

```python
for i in range(pasos):
  if i == 0:
    Es[i] = newton(lambda E: E - e*np.sin(E) - M, M)
    fs[i] = anomalia_verdadera(Es[i])
  else:
    Ms[i] = Ms[i - 1] + 6 * n
    Es[i] = newton(lambda E: E - e*np.sin(E) - Ms[i], Ms[i])
    fs[i] = anomalia_verdadera(Es[i])
```

Definimos una función que calcule las componentes $x, y, z$ de la posición de Apophis en el sistema del observador mediante las ecuaciones:

$x = r[cos(\Omega)cos(\omega + f) - cos(i)sin(\Omega)sin(\omega+f)]$

$y = r[sin(\Omega)cos(\omega + f) + cos(i)cos(\Omega)sin(\omega+f)]$

$z = r[cos(f)sin(\omega)sin(i) + sin(f)cos(\omega)sin(i)]$

Y tomando:

$r(f) = \frac{p}{1+ecos(f)}$

```python
def posicion(p, e, i, nodo, argumento, f): #nodo: Omega. argumento: omega

  radio = p / (1 + e * np.cos(f))

  x = (radio * (np.cos(nodo) * np.cos(argumento + f) \
      - np.cos(i) * np.sin(nodo) * np.sin(argumento + f)))
  y = (radio * (np.sin(nodo) * np.cos(argumento + f) \
      + np.cos(i) * np.cos(nodo) * np.sin(argumento + f)))
  z = (radio * (np.cos(f) * np.sin(argumento) * np.sin(i) \
      + np.sin(f) * np.cos(argumento) * np.sin(i)))

  return x, y, z
```

Definimos los arrays para guardar las componentes $x, y, z$ de la posición de Apophis en el sistema del observador

```python
xs = np.zeros(pasos)
ys = np.zeros(pasos)
zs = np.zeros(pasos)
```

Calculamos las componentes $x,y,z$ para cada tiempo usando la función definida anteriormente

```python
for t, f in enumerate(fs):
  xs[t], ys[t], zs[t] = posicion(p, e, incl, Omega, omega, f)
```

```python
mu = G2 * (pc.constantes.mu_earth / (pc. constantes.G * UM))

r = np.linalg.norm(rs_relativos)
v = np.linalg.norm(vs_relativos)
```

```python
h_vec = np.cross(rs_relativos, vs_relativos)
h = np.linalg.norm(h_vec)

e_vec = np.cross(vs_relativos,h_vec)/mu - rs_relativos / r
e = np.linalg.norm(e_vec)
```
